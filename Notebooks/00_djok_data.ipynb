```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/atp_tennis.csv")
```

```python
df_true = df[(df["Player_1"] == "Djokovic N.") | (df["Player_2"] == "Djokovic N.")]
```

```python
# Checking whether every row is a match with Djokovic
has_dj = (df_true[["Player_1", "Player_2"]] == "Djokovic N.").any(axis=1)
print(has_dj.all())
```
