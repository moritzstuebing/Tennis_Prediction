# Random Forests Classification

In this notebook, we will develop a random forests classifier using scikit-learn to classify our data according to whether Djokovic won the match or not. In this notebook, we will use scikit-learn as much as possible while still preserving the ability to compare the performance of the random forests model to other the logistic regression and neural networks we have applied to the same task.

## 1. Imports and Data Preparations

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from utils import shuffle_data_df
```

```python
df = pd.read_csv("../Data/full_data.csv")
```

```python
df.head()
```

We will want to make sure that the data is shuffled in the exact same way as in the logistic regression notebook. The shuff_data function has a built in seed value so calling it normally on our data will do the job.

```python
# Shuffling using same seed as log reg notebook
df_shuff = shuffle_data_df(df)
```

```python
# Extracting features and targets
y = df_shuff['dj_win']
X = df_shuff.iloc[:, df.columns != 'dj_win']
```

```python
# One hot encoding
enc = OneHotEncoder(sparse_output=False)
enc.fit(X.iloc[:,:3])
X_dummies = pd.DataFrame(
    enc.transform(X.iloc[:, :3]),
    columns=enc.get_feature_names_out(),
    index=X.index,
)
X_oh = pd.concat([X.iloc[:, 3:], X_dummies], axis=1)
```

```python
# Shuffle = False to ensure continuity with log reg notebook
X_train, X_test, y_train, y_test = train_test_split(
    X_oh, y, test_size=0.2, shuffle=False)
```

## 2. Fitting Random Forests Model

```python
rf = RandomForestClassifier()
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
rf.score(X_test, y_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
features_1 = pd.DataFrame(rf.feature_importances_, index = X_oh.columns)
```

```python
features_1
```

The table above shows the importance of individual features in the classification of the data. We can see that the most important features are dj_pts, dj_odds, opp_pts, opp_rank, opp_odds, and age to a slightly lesser extent. This is all expected since these clearly give good indication of how tough the opponent is, and odds essentially give a bookmaker-calibrated probability of Djokovic winning the match which is already very good signal for the end result of the match. Interestingly, dj_rank is not very important which is probably because he was ranked highly for most of his career, so there is not much signal contained in the differences of his own rank.

## 3. Modified Model

Now we will train a second random forest with some hyperparameters. n_estimators is the number of trees, increasing this will decrease the variance of the prediction, cancelling out individual error in trees and smoothing noise. The criterion parameter is the formula used to decide the best split to take at each node, i.e for all features chosen in the random choice, for all possible split points, the criterion increase/decrease in the criterion of choosing this split is calculated and the optimal one chosen. min_samples_split gives the minimum samples required to split an internal node, once the samples in a node are below this, no more splits will be computed on this node.

```python
# hyperparameters
rf2 = RandomForestClassifier(n_estimators = 1000,
                             criterion = 'gini',
                             min_samples_split = 10,
                             random_state = 42)
```

```python
rf2.fit(X_train, y_train)
```

```python
y_pred = rf2.predict(X_test)
```

```python
rf2.score(X_test, y_test)
```

```python
print(classification_report(y_test, y_pred))
```

From the classification report, we can see that the model has good precision, meaning that when it does choose $0$/$1$ it is correct $75$/$84$% of the time. However, the recall gives us an idea of what the model is actually learning. We can see that for the label $1$, the model has a $99$% recall rate, meaning that out of all of the true positives in the data, it catches almost all of them. But, the recall rate for label $0$ is $13$%, which is very low, meaning that out of all $0$ labels in the data, it only catches $13$% of them. This indicates that the model has learnt to just guess "win" most of the time, because Djokovic has a high win rate, meaning doing so will lead to a relatively high accuracy. Guessing "win" most of the time will of course mean you catch most of the true wins, and a small amount of the losses, as the recall indicates. We can see in the support column that there are $205" wins and "45" losses. Hence, just choosing win for every match gives an accuracy of $82$%, which the model already sees as good.

## 4. Handling Class Imbalance

We will now train a random forest with even loss weights for each of the classes. This will encourage the model to actually attempt to predict losses more.

```python
rf3 = RandomForestClassifier(n_estimators = 1000,
                             criterion = 'gini',
                             min_samples_split = 10,
                             random_state = 42,
                             class_weight='balanced')
```

```python
rf3.fit(X_train, y_train)
```

```python
features_3 = pd.DataFrame(rf3.feature_importances_, index = X_oh.columns)
```

```python
features_3
```

```python
y_pred = rf3.predict(X_test)
```

```python
rf3.score(X_test, y_test)
```

```python
print(classification_report(y_test, y_pred))
```

We can see now that, although the score has decreased, the recall of the losses has improved, meaning that the model is actively trying to catch the losses as well as the wins instead of just guessing win most of the time.

## 5. Discussion

In this notebook, we trained a random forest on the full dataset, including all features. We saw that the model can achieve $\approx 84 \%$ accuracy. This was achieved by the models mainly choosing to predict win, because since Djokovic's win rate is very high the model will seem to do well even though they are just guessing win every time. Due to this, we introduced some class imbalance handling, where we modified the weights for datapoints from the different classes to force the model to attempt to extract loss signal aswell. Although this deceased accuracy from $\approx 84\%$ to $\approx 80\%$, the negative recall (number of losses that the model correctly classified) increased from $\approx 13-18\%$ up to $\approx 60 \%$, a substantial improvement.

The logistic regression and this notebook bring up some interesting points. We saw that the odds are by far the most important feature for classification, since they give a bookmaker-algorithm calculated probability of Djokovic winning the match. These are already well-calibrated and will predict most of the signal in the data, including features that are difficult to engineer like current news (e.g "he tweaked something in practice"). Hence, trying to classify while using the odds as a feature already uses the best possible feature of win and loss indication. When adding further features than just the odds, we are trying to predict the signal that the market missed, which is a difficult task. Furthermore, the model learns that odds are a very good indicator for our classification and contain the most signal by far, so just learns to predict using this. Any of the other features, like age, rank, 5SMA, are given much less importance. The accuracies our model achieve are essentially just a reflection of how good the bookmaker odds are at capturing the signal.

This is also an artefact of the player we have chosen. Djokovic wins most of his matches, and we have seen that some models default to just guessing "win" for every match to attain a relatively good accuracy. The losses he does have are often upsets, and there is not much signal to predict these more random events. Hence, when choosing to pursure hard classification, as we have seen the model almost always just pursues the win rate. The better thing to try and predict may be the probability of winning / losing.
