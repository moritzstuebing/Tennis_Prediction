# Random Forests Classification

In this notebook, we will develop a random forests classifier using scikit-learn to classify our data according to whether Djokovic won the match or not. In this notebook, we will use scikit-learn as much as possible while still preserving the ability to compare the performance of the random forests classifier to the logistic regression classifier from the previous notbeook.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from utils import shuffle_data, standardise_cols
```
